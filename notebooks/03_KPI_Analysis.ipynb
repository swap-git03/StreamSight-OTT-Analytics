## Notebook 03: Executive Business KPI Analysis

### Objective:
Calculate core Business KPIs: Total Watch Hours, Overall Completion Rate (%), Average Content Rating, Subscription Revenue Breakdown, and Payment Success Rates.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../dataset/cleaned_streamsight_data.csv')

# 1. Total Watch Hours
total_watch_hours = df['watch_hours'].sum()

# 2. Overall Completion Rate
completion_rate = (df['completed'] == 'Yes').mean() * 100

# 3. Average Rating
avg_rating = df['rating'].mean()

# 4. Payment Failure Rate
payment_failure_rate = (df['payment_status'] == 'Failed').mean() * 100

print('================ EXECUTIVE KPI SUMMARY ================')
print(f'Total Watch Hours      : {total_watch_hours:,.2f} Hours')
print(f'Overall Completion Rate: {completion_rate:.2f}%')
print(f'Average Content Rating : {avg_rating:.2f} / 5.0')
print(f'Payment Failure Rate   : {payment_failure_rate:.2f}%')
print('=======================================================')
```

### KPI Breakdown by Subscription Plan

```python
plan_kpis = df.groupby('subscription_plan').agg(
    total_users=('user_id', 'nunique'),
    total_sessions=('watch_id', 'count'),
    total_watch_hours=('watch_hours', 'sum'),
    completion_rate=('completed', lambda x: (x == 'Yes').mean() * 100),
    avg_rating=('rating', 'mean')
).reset_index()

display(plan_kpis.round(2))
```
