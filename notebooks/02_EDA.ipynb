# StreamSight – OTT Business Intelligence Platform
## Notebook 02: Exploratory Data Analysis (EDA)

### Objective:
Perform univariate, bivariate, and multivariate analysis on the cleaned master dataset to understand user demographics, watch duration distributions, genre trends, and device usage.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style='darkgrid')
df = pd.read_csv('../dataset/cleaned_streamsight_data.csv')
df.head()
```

### 1. Summary Statistics & Data Profiles

```python
print('Numerical Features Summary:')
display(df[['user_age', 'minutes_watched', 'watch_hours', 'rating', 'release_year']].describe())

print('Categorical Features Unique Counts:')
display(df[['gender', 'country', 'subscription_plan', 'content_type', 'genre', 'device', 'completed', 'payment_status']].nunique())
```

### 2. User Demographics & Subscription Tier Distribution

```python
# Subscription plan distribution
plan_counts = df['subscription_plan'].value_counts()
print('Subscription Plan Distribution:')
print(plan_counts)

# Age group distribution
age_counts = df['age_group'].value_counts()
print('\nAge Group Distribution:')
print(age_counts)
```

### 3. GroupBy & Pivot Table Analysis: Watch Duration by Genre and Content Type

```python
# Pivot table: Mean Watch Hours by Genre & Content Type
pivot_genre = df.pivot_table(
    index='genre',
    columns='content_type',
    values='watch_hours',
    aggfunc=['count', 'mean', 'sum']
)
display(pivot_genre)
```

### 4. Bivariate Analysis: Device Dominance vs Completion Rate

```python
device_completion = pd.crosstab(
    df['device'],
    df['completed'],
    normalize='index'
) * 100

print('Completion Rate (%) by Device:')
display(device_completion.round(2))
```
