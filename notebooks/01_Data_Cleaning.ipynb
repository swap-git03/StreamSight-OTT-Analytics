# StreamSight – OTT Business Intelligence Platform
## Notebook 01: Data Cleaning & Feature Engineering

### Objective:
1. Load raw streaming event logs (`StreamSight_OTT_Analytics_Dataset.csv`).
2. Perform data quality audit (check missing values, duplicates, out-of-range values).
3. Sanitize data types and standardize categorical fields.
4. Engineer business features (`watch_hours`, `age_group`, `watch_month_name`, `is_weekend`, `content_age_years`).
5. Export master cleaned dataset and 3NF normalized tables (`users.csv`, `content.csv`, `watch_sessions.csv`, `payments.csv`).

```python
import os
import sys
import pandas as pd
import numpy as np

# Import custom helper module from python directory
sys.path.append(os.path.abspath('../python'))
from helper import load_dataset, audit_data_quality, clean_string_fields, convert_data_types, engineer_features, extract_relational_tables
```

### Step 1: Load Dataset & Audit Data Quality

```python
raw_path = '../dataset/StreamSight_OTT_Analytics_Dataset.csv'
df_raw = load_dataset(raw_path)
audit_results = audit_data_quality(df_raw)
df_raw.head()
```

### Step 2: Handle Duplicates and Standardize Text

```python
df_clean = df_raw.drop_duplicates().copy()
df_clean = clean_string_fields(df_clean)
df_clean = convert_data_types(df_clean)
print('Data types after conversion:')
print(df_clean.dtypes)
```

### Step 3: Feature Engineering
Generating metrics like `watch_hours`, `age_group`, date parts, and `content_age_years`.

```python
df_clean = engineer_features(df_clean)
df_clean[['minutes_watched', 'watch_hours', 'user_age', 'age_group', 'watch_date', 'watch_month_name', 'is_weekend', 'content_age_years']].head()
```

### Step 4: Export Master Clean Dataset and 3NF Relational Tables

```python
users_df, content_df, watch_sessions_df, payments_df = extract_relational_tables(df_clean)

print(f'Users count: {len(users_df)}')
print(f'Content count: {len(content_df)}')
print(f'Watch sessions count: {len(watch_sessions_df)}')
print(f'Payments count: {len(payments_df)}')

# Export files
df_clean.to_csv('../dataset/cleaned_streamsight_data.csv', index=False)
users_df.to_csv('../dataset/users.csv', index=False)
content_df.to_csv('../dataset/content.csv', index=False)
watch_sessions_df.to_csv('../dataset/watch_sessions.csv', index=False)
payments_df.to_csv('../dataset/payments.csv', index=False)
print('All clean datasets successfully exported to ../dataset/')
```
