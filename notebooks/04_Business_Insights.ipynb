## Notebook 04: Strategic Business Insights & Recommendations

### Objective:
Deep dive into high-value business insights: payment failure correlation with user churn, country-level viewing density, genre popularity trends, and device-level streaming efficiency.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../dataset/cleaned_streamsight_data.csv')

# 1. Country Level Analysis
country_metrics = df.groupby('country').agg(
    unique_users=('user_id', 'nunique'),
    total_sessions=('watch_id', 'count'),
    total_hours=('watch_hours', 'sum'),
    avg_rating=('rating', 'mean')
).sort_values('total_hours', ascending=False)

print('Country Viewing Performance:')
display(country_metrics.round(2))
```

### 2. Payment Status vs Completion Rate Insight

```python
payment_insight = df.groupby(['payment_status', 'payment_method']).agg(
    session_count=('watch_id', 'count'),
    completion_rate=('completed', lambda x: (x == 'Yes').mean() * 100),
    avg_watch_minutes=('minutes_watched', 'mean')
).reset_index()

print('Payment Method & Status Impact:')
display(payment_insight.round(2))
```
